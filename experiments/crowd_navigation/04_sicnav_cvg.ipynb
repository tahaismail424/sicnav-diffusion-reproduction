# SICNav-CVG
SICNav-CVG runs the interactive ORCA lower-level human model inside acados. Human intended goals are estimated by projecting current velocity (`CVG`), rather than by a neural trajectory forecaster. This isolates interactive planning from diffusion prediction.

```python
from pathlib import Path
import sys
here = Path.cwd().resolve()
ROOT = next(p for p in (here, *here.parents) if (p / 'experiments/crowd_navigation/navigation.py').exists())
EXP = ROOT / 'experiments/crowd_navigation'
sys.path.insert(0, str(EXP))
from navigation import dependency_report, plot_episode, run_navigation_suite, summarize
dependency_report('sicnav_cvg')
```

The dependency report must show `casadi`, `acados_template`, and `ACADOS_SOURCE_DIR` as ready. The first run generates and compiles an acados problem cache for this controller geometry.

```python
SCENARIO = 'hallway_bottleneck'
HUMAN_COUNT = 3
CASE = 0
one, episodes = run_navigation_suite('sicnav_cvg', cases=[CASE], scenario=SCENARIO, human_count=HUMAN_COUNT, device='auto')
display(one)
plot_episode(episodes[CASE], title=f'SICNav-CVG: {SCENARIO}, N={HUMAN_COUNT}, case={CASE}');
```

```python
results, _ = run_navigation_suite('sicnav_cvg', cases=range(5), scenario=SCENARIO, human_count=HUMAN_COUNT, device='auto', output_dir=EXP/'outputs/policy_benchmark')
summarize(results)
```
