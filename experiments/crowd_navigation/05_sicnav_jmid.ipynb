## Closed-loop episode
Keep `starts_moving >= 7`: JMID needs enough simulator history before its first forecast. The released controller and checkpoint are configured for three humans by default; validate that setting before changing scene cardinality.

```python
SCENARIO = 'hallway_bottleneck'
HUMAN_COUNT = 3
CASE = 0
one, episodes = run_navigation_suite('sicnav_jmid', cases=[CASE], scenario=SCENARIO, human_count=HUMAN_COUNT, starts_moving=10, device='auto')
display(one)
plot_episode(episodes[CASE], title=f'SICNav-JMID: {SCENARIO}, N={HUMAN_COUNT}, case={CASE}');
```

```python
results, _ = run_navigation_suite('sicnav_jmid', cases=range(5), scenario=SCENARIO, human_count=HUMAN_COUNT, starts_moving=10, device='auto', output_dir=EXP/'outputs/policy_benchmark')
summarize(results)
```
